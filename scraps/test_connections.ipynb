# Test Model Endpoints

Testing available models and completion endpoints

```python
import requests
from openai import OpenAI
import json
```

## Check Available Models

```python
# Port 12302 - Qwen3-30B endpoint
client = OpenAI(
    base_url='http://96.245.177.243:12302/v1',
	# base_url='http://0.0.0.0:12302/v1',
    api_key="none"
)

print("Available models on port 12302:")
models = client.models.list()
for model in models.data:
    print(f"  - {model.id}")
```

## Test Completion with requests (equivalent to curl)

```python
# Equivalent to the curl command
url = 'http://96.245.177.243:12302/v1/completions'
headers = {'Content-Type': 'application/json'}
data = {
    'model': 'cpatonn/Qwen3-30B-A3B-Instruct-2507-AWQ-8bit',
    'prompt': 'San Francisco is a',
    'max_tokens': 7,
    'temperature': 0
}

response = requests.post(url, headers=headers, json=data)

print(f"Status: {response.status_code}")
print("\nResponse:")
print(json.dumps(response.json(), indent=2))
```

## Test with OpenAI Client (Chat Completions)

```python
# Using chat completions instead of completions
prompt = """Answer this movie question with a simple list.

Question: the films acted by [Shaun White] were in which genres

Instructions:
- List ONLY answers you are confident about
- One per line

Your answer:"""

completion = client.chat.completions.create(
    model="cpatonn/Qwen3-30B-A3B-Instruct-2507-AWQ-8bit",
    max_tokens=100,
    messages=[
        {"role": "system", "content": "You are a helpful assistant."},
        {"role": "user", "content": prompt}
    ]
)

print("Model response:")
print(completion.choices[0].message.content)
print("\nFull response:")
print(completion)
```

## Test Different Prompts

```python
# Test with different prompts
test_prompts = [
    "San Francisco is a",
    "The capital of France is",
    "Python is a programming language that"
]

for prompt in test_prompts:
    response = requests.post(
        # 'http://96.245.177.243:12302/v1/completions',
		url = 'http://0.0.0.0:12302/v1/completions',
        headers={'Content-Type': 'application/json'},
        json={
            # 'model': 'cpatonn/Qwen3-30B-A3B-Instruct-2507-AWQ-8bit',
			'model': 'Qwen3-30B-A3B-Instruct-2507',
            'prompt': prompt,
            'max_tokens': 20,
            'temperature': 0
        }
    )
    print(response)
    result = response.json()
    completion_text = result['choices'][0]['text']
    
    print(f"Prompt: {prompt}")
    print(f"Completion: {completion_text}")
    print("-" * 60)
```
